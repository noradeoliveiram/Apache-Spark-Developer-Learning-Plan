## Methods

A method is a special function that acts on an object that we call like this: **object.method(args)**

We define a method:
1. We nest the definition of the method within the class definition.
2. We must specify a parameter named _self_, followed by any additional parameters.

Definition:

```
class ClassName():

    def method_name(self, args):
        method code
```

````python
class UpdatedDog():

    def return_name(self, name):
        return(f"Name: {name}")
````

````python
my_updated_dog = UpdatedDog()
my_updated_dog.return_name("Odie")
````

A _method_ differs from a _function_ in that it can act on the object it is called on. The _method_ needs to be able to reference the object that called it. That's what _self_ refers to.

When we call **object.method(self, args)** the object itself is passed to the self parameter automatically by Python.

````python
class DogWithSelf():

    def print_self(self):
        print(self)


my_dog_with_self = DogWithSelf()
print(my_dog_with_self)

my_dog_with_self.print_self()
````

## Attributes

In our class, we need some way to store data. When a variable is stored in a class, it is called an attribute. Attributes are just variables that are defined for each instance of an object. Every instance will have the same named attributes byt they are normally set to different values.

```
class ClassName():

    def __init__(self, arg):
        self.arg = arg
```

Python provides a special method called **__init__(self)** that is automatically called when our object is initialized. This is often referred to as the _constructor method_ for the class since it constructs the class's attributes.

````python
class DogWithAttributes():

    def __init__(self, name, color):
        print("This ran automatically")
        self.name = name
        self.color = color


my_dog_with_attributes = DogWithAttributes("Odie", "Brown")
````

````python
# We can access the attribute of af an object similar to how we accessed methods but without parentheses.
my_dog_with_attributes.name
````

In a method definition we can access an attribute using **self.attribute_name**, since self refers to the object that calls the method.

````python
class DogWithAttributesAndMethod():

    def __init__(self, name, color):
        self.name = name
        self.color = color

    def return_name(self):
        return self.name
    
    
my_dog = DogWithAttributesAndMethod("Odie", "Brown")
my_dog.return_name()
````

````python
class DogWithAttributesAndMethods():

    def __init__(self, name, color):
        self.name = name
        self.color = color

    def return_name(self):
        return self.name
    
    def update_name(self, new_name):
        self.name = new_name
        

my_dog = DogWithAttributesAndMethods("Odie", "Brown")
print(f"Here is my dog's name before I update it: {my_dog.return_name()}")

my_dog.update_name("Tami")
print(f"Here is my dog's name after I update it: {my_dog.return_name()}")
````

Classes can have many methods and attributes. They can also access the attributes of another class.

````python
class DogFinal():

    def __init__(self, name_str, color_str):
        self.name = name_str
        self.color = color_str

    def return_name(self):
        return self.name
    
    def update_name(self, new_name):
        self.name = new_name

    def return_both_names(self, other_dog_object):
        return self.name + " and " + other_dog_object.name


dog_1 = DogFinal("Odie", "Brown")
dog_2 = DogFinal("Tami", "Black")

dog_1.return_both_names(dog_2)
````
