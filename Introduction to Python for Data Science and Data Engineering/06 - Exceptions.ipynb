## Exceptions

If we have properly formatted code that Python knows how to run, we might still encounter errors as he code is executed. Errors that arise like this as the code is executed are known as Exceptions. They indicate that, while Python understood what we were trying to do, there is a problem.

```
1 / 0
```

### Exception handling

When Python encounters a _try_ statement, it will first try to run the code in the _try_ code block. If it encounters an exception, instead of exiting and erroring, it will instead run the code block under _except_.

```
try:
  code block
except:
  code block
```

If we want to only handle a cetain exception we can write the exception after the _except_ keyword:

```
try:
  code block
except ExceptionName:
  code block
```

If we want to handle multiple specific exceptions, we can write a sequence of exceptions separated by commas inside parentheses.

````python
try:
    1 / 0
except:
    print("Exception handled")
````

````python
try:
    1 / 0   # ZeroDivisionError exception
except ZeroDivisionError:
    print("Exception handled")
````

````python
try:
    # 1 / 0   # ZeroDivisionError exception
    print(undefined_variable)   # NameError exception
except (ZeroDivisionError, NameError):
    print("Exception handled")
````

### Assertion error

One very useful exception is _AssertionError_. We can raise _AssertionErrors_ using assert.

Definition: assert boolean expression, optional message

When Python executes this statement, it evaluates the boolean expression first. If it is _True_, Python does nothing and moves on. If it is _False_, then it throws an _AssertionError_ with the optional message, if provided.

````python
assert 1 == 1
assert 1 == 2, "That is not true"
````
