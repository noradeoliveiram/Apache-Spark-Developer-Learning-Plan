
# Data types and variables

```python
# Integer

2 * 3 + 5 - 1
```

```python
# Float

1.2 * 2.3 + 5.5

```

```python
# String

print("Hello" + "123")
print("Hello" + " " + "123")

# "Hello" + 123 ---> Error: can only concatenate str (not "int") to str
```

```python
# Boolean

print(True or False)
print(True and False)
print(not False)

```

```python
# Type

print(type(1))
print(type(1.))
print(type("Hello"))
print(type(True))
```

```python
# Variables

a = 3
b = 2
c = a * b

c
```

```python
b = 4
c
```

```python
# Python is a weakly typed language

b = "Hellow World"
b = 1

b
```

```python
# Print statements

a = 1
b = "Hello"

a # This line isn't printed because it's not the last line of code
b
```

```python
print(a)
print(b)
```

```python
# f-string formatting

name = "Nora"
age = 22

print(f"My name is {name} and i am {age} years old")
```
