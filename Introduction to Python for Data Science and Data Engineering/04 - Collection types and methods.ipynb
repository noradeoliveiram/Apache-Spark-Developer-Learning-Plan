## Objects

An object is an instance of a specific data type

## Methods

Methods are functions provided by a data type that we can call on objects of that type. They act on an object we call them with and allow us to use more functionality provided by that data type.

```
object.method.name(parameters)
```

### String methods

````python
greeting = "hello"

print(greeting.upper())
print(greeting)
````

### In-place methods

Methods are functions that act on objects, and can either perform operations in-place (modify the underlying object it was called upon) or return a new object.

A list is just an ordered sequence of items.

Definition: [item1, item2, item3...]

The items may be of any type, though in practice you'll usually create lists where all of the values are of the same type.

````python
breakfast_list = ["pancakes", "eggs", "waffles"]
breakfast_list
````

**List indexing**

Often, we want to reference a specific item or items in a list. This is calle list indexing.

Lists provide an operation to get the item at a certain index in the list like this:
```
list_name[index]
```

- In Python indices start from 0.
- Negative indexing, which starts counting from right lo left, starting from -1.
- Range of indices: returns a list of the values starting at _start_ and up to but not including _stop_. If we don´t provide a start index, Python assumes we start at the beginning. If we don't provide a stop index, Python assumes we stop at the end.
```
list_name[start:stop] 
```

````python
print(breakfast_list[2])

print(breakfast_list[-1])

print(breakfast_list[1:3])

print(breakfast_list[:3])

print(breakfast_list[1:])
````

We can change the value of an index in a list:

````python
print(breakfast_list)

breakfast_list[0] = "sausage"

print(breakfast_list)
````

**In:** to check if an element is in a given list. This is a boolean operation:

````python
"waffles" in breakfast_list
````

### Dictionaries

A Dictionary is a sequence of key-value pairs.

Definition: {key1: value1, key2: value2, ...}

The keys and values can all be of any type. However, because each key maps to a value, it is important that all keys are unique.

````python
breakfast_dictionary = {"pancakes": 1, "eggs": 2, "waffles": 3}
breakfast_dictionary
````

#### Dictionary methods

Dictionaries provide the method **dictionary_object.get()** to get the value in the dictionary for the given argument.

Alternatively, we can use the syntax **dictionary_object[key]**.

````python
print(breakfast_dictionary.get("waffles"))
print(breakfast_dictionary["waffles"])
````

We can update a dictionary similarly to a list by assingning **dictionary_object[key]** to be something.

If the key is present, it overwrites the current value. If not, it creates a new key-value pair.

````python
print(breakfast_dictionary)

breakfast_dictionary["waffles"] += 1
breakfast_dictionary["yogurt"] = 1

print(breakfast_dictionary)
````

In order to determinate if a key is in a dictionary, we can use the method **dictonary_name.keys()**. This returns a list of the keys in the dictionary.

Similar to lists, we can use **in** to see if out key is in the dictionary.

````python
print(breakfast_dictionary.keys())

print("bacon" in breakfast_dictionary.keys())
````
