
# PANDAS

Pandas is a popular Python library among data scientists with high-performing, easy-to-use data structures and data analysis tools.

```python
import pandas as pd
```

## DataFrame

**Pandas** is a library that provides data types and functions that allows us to do rigorous, programmatic data analysis.

The core pandas data type is called **DataFrame**.

A **DataFrame** is a two dimensional table of named rows and columns, similar to a SQL table.

The **DataFrame** class has a _data_ attribute for the data in the table that we have to define when we instantiate a **DataFrame** object.

We want to make the following table into a **DataFrame**:

| Name | Age | Job |
|:---:|:---:|:---|
| John | 30 | Journalist |
| Mary | 30 | Programmer |
| Abe | 40 | Chef |

```python
data = [["John", 30, "Journalist"], ["Mary", 30, "Programmer"], ["Abe", 40, "Chef"]]

df = pd.DataFrame(data = data)
df
```

### Adding column names

To specify the column names we want, DataFrames has another attribute: **columns**.

```python
cols = ["Name", "Age", "Job"]

df = pd.DataFrame(data = data, columns = cols)
df
```

## Series

**Series** are the other main data type that **pandas** provides.

A **Series** is just one column of the **DataFrames**.

We can select a **Series** in two ways:
1. df["column_name"]
2. df.column_name

```python
df["Age"]
```

```python
df.Age
```

### dtypes

If you look at the `Series` object above you can see `dtype: int64`.

In **pandas** `dtypes`, which is an abbreviation for data types, refers to the type of the values in the column.

The **dtype** of a column determines what functionality we can do with it.

For example, we can take the average of a numeric column, but not a non-numeric one.

The `dtypes` of columns are specific to **pandas**, but the most common ones are very similar to the built-in Python types:
1. `object` is just text, which is similar to strings
2. `int64` are integers
3. `float64` are float
4. `bool` are boolean

We can view the `dtypes` of every column in our `DataFrame` by accessing its `dtypes` attribute:

```python
df.dtypes
```

### Series operations

We can use operations on `Series` of a certain dtype that are similar to the operations we can do with the dtype's similar built-in counterpart.

These operations act element-wise.

For example, we can add `int64` `Series` similarly to how we can add integer values in Python.


```python
df["Age"] + df["Age"]
```

```python
df["Age"] * 3 - 1
```

### Selecting a value from a `Series`

```python
# First element

df["Age"][0]
```

### Selecting a Subset of columns

We can select a subset of columns like this:

**df[[col1, col2, col3, ...]]**

```python
df[["Name", "Age"]]
```
