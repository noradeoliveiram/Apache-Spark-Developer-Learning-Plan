## Cloud

Managing an on-premise system is difficult, expensive and scales poorly. An popular alternative is to rent storage and computer power from cloud providers (Amazon, Microsoft, Google...).

In this situation, a user can simply access data and compute via a web browser or other application, while the actual data and computation are being stored and ran in large warehouse of machines called a data center managed by a company. This is referred to as a cloud-based setting.

It is much less expensive and easier to use cloud storage because you don't have to create or manage your own data center. It also allows for easy scaling: just buy as much storage and computer power as you need at the moment and turn it off when you finish.

### Virtual machines

In a cloud based setting we use computers managed by cloud providers to run code and store data.

We are able to run code this way by using virtual machines on those computers.

A virtual machine separates the CPU, memory, networking, and disk storage from other virtual machines on the same computer.

By renting virtual machines on cloud computers, we can use the resources those computers provide without worrying about sharing information with other users also renting virtual machines.


### Cloud storage

Cloud providers offer ways to store data on the cloud easily. These services use computers and software that are specialized for storing data in a reliable way that can scale well.

One type of storage offered by cloud providers is **object storage**, which can store any type of data including text, images, videos, and other binary data. Some examples of cloud object storage are:

- Amazon Simple Storage Service (Amazon S3)
- Microsoft's Azure Blob storage
- Google Cloud Storage

Cloud providers also offer services to store and manage relational databases — such as MySQL, PostgreSQL, and Microsoft SQL Server — and key-value stores or other "NoSQL" databases — such as Amazon DynamoDB, Azure Cosmos DB, and Google Cloud Bigtable.

### Apache Spark

A single computer usually has the memory and computational power to perform calculations on data sets up to the size of a few gigabytes or less. Data sets larger than that either can't fit into the memory of a single computer or take an unacceptably long time for a single computer to process. For these types of "big data" use cases, we need a system that can split a large data set into smaller subsets — often referred to as partitions — and then distribute the processing of these data partitions across a number of computers.

Apache Spark is an open-source data processing engine that manages distributed processing of large data sets.

For example, let's say that we have a large data set and we want to calculate various statistics for some of its numeric columns. With Apache Spark, our program only needs to specify the data set to read and the statistics that we want calculated. We can then run the program on a set of computers that have been configured to serve as an Apache Spark cluster. When we run it, Spark automatically:

- determines how to divide the data set into partitions,
- assigns those partitions to the various computers of the cluster with instructions for calculating per-partition statistics, and
- finally collects those per-partitions statistics and calculates the final results we requested.

Spark was created originally as a research project at the University of California Berkeley. In 2013, the project was donated to the Apache Software Foundation. That same year the creators of Spark founded Databricks.

Databricks, in general, uses Apache Spark as the computation engine for the platform. Databricks provides simple management tools for running Spark clusters composed of cloud-provided virtual machines to process the data you have in cloud object storage and other systems.

### Databricks File System - DBFS

Databricks File System (DBFS) is a distributed file system mounted into a Databricks workspace and available on Databricks clusters. DBFS is an abstraction on top of scalable object storage and offers the following benefits:

- Allows you to mount storage objects so that you can seamlessly access data without requiring credentials.
- Allows you to interact with object storage using directory and file semantics instead of storage URLs.
- Persists files to object storage, so you won't lose data after you terminate a cluster.

Mounting object storage to DBFS allows you to access objects in object storage as if they were on the local file system. Let's take a look at our currently mounted data by calling **%fs mounts**, or alternatively, **dbutils.fs.mounts()**.
